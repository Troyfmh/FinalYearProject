```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Draw import IPythonConsole  # Enables inline molecule display
import glob

# Folder path (adjust if necessary)
folder_path = r"C:\Project_code\15_molecules\*.sdf"

# Get all SDF files in the folder
sdf_files = glob.glob(folder_path)

# Loop through and display molecules
for sdf_file in sdf_files:
    suppl = Chem.SDMolSupplier(sdf_file)
    
    for mol in suppl:
        if mol is not None:
            print(f"Displaying: {sdf_file}")
            img = Draw.MolToImage(mol)  # Displays all 15 molecules as indiviudal pictures
            img.show()
```
